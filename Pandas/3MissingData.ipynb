Missing Data

```python
import numpy as np
import pandas as pd
```

1.Finding Missing Data

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna() # give me true if there is any null value in the data
```

```python
df.isna().sum() # gives the number of null values in every column
```

```python
df.isna().any() # give us true if every column has null values
```

2.Remove null values

```python
df
```

```python
# ye work karta hai rows ke basis pe 
df.dropna() # delete every row which has null value 
```

```python
df
```

```python
df.dropna(thresh=3) # we can give thresh so if there is 3 number of non-null values it give us data
```

3.Filling the missing data

```python
df
```

```python
df.fillna(0) # fill null values with 0
```

```python
# we want to fill the null values in each column with different numbers
values={"A":30,"B":100,"C":300,"D":400}
df.fillna(value=values)
```

```python
# fill data with mean
df.fillna(df.mean())
```
