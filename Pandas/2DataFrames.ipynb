```python
import numpy as np
import pandas as pd
```

1.Creating DataFrame

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
} # data

df=pd.DataFrame(data)
df
```

```python
# using a list where we dont have any headers
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
columns=["Name","Age","City","Salary"]
df2=pd.DataFrame(data_list,columns=columns)
df2
```

3.Creating a new column


```python
df2["Designation"]=["Doctor","Eng","Doctor","Eng"]
df2
```

4.Remove a column

```python
df2.drop("Designation",axis=1) # drop se main data se remove nahi hota hai vo bas copy dikhata hai ki aisa dikhega apka data 
```

```python
# to agar hame main data se bhi hatana hai to hum use karenge inplace=True
df2.drop("Designation",axis=1,inplace=True)

```

```python
df2
```

```python
# delete multiple columns
df2.drop(["City","Salary"],axis=1,inplace=True)
```

```python
df2
```

```python
# we can delete a row also just give axis =0
df2.drop(0,axis=0)
```

6.Selecting subsets of rows and columns

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
} # data

df3=pd.DataFrame(data)
df3
```

```python
df3.loc[[0,1]][["City","Salary"]] # if we want to select specific portion
```

```python
df3.loc[[2,3]][["Name","Age"]]
```

7.Conditional Selection


```python
df3
```

```python
# i only want to see those people whose age is above 30
df3[df3["Age"]>30]
```

```python
# i onlt want people whose age is above 30 and their city must be paris
df3[(df3["Age"]>30) & (df3["City"]=="Paris")]
```
