1.Numpy Indexing And Slicing of Vectors 


```python
arr1=np.arange(11,21)
print(arr1)
```

```python
arr1[1:5]
```

```python
arr1[2:9:2]
```

2.Numpy Indexing and Slicing of Matrix

```python
arr2=np.arange(1,31).reshape(6,5)
print(arr2)
```

```python
print(arr2[5,4])   # Indexing
```

```python
print(arr2[0:2,1:3])
```

```python
print(arr2[3:5,3:5])
```

3.Boolean Indexing

```python
arr3=np.arange(11,21)
print(arr3)
```

```python
bool_Index=arr3%2==0
print(bool_Index)
```

```python
arr4=arr3[bool_Index]
print(arr4)
```
